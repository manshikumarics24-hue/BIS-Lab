```python

import random
import math

# 1. Restaurant and customer locations
locations = {
    "Restaurant": (0, 0),
    "Customer1": (2, 5),
    "Customer2": (5, 2),
    "Customer3": (6, 7),
    "Customer4": (8, 3)
}

names = list(locations.keys())
n = len(names)

# 2. Calculate distances
dist = [[0] * n for _ in range(n)]

for i in range(n):
    for j in range(n):
        if i != j:
            x1, y1 = locations[names[i]]
            x2, y2 = locations[names[j]]

            dist[i][j] = math.sqrt(
                (x2-x1)**2 + (y2-y1)**2
            )

# 3. Initialize parameters
num_ants = 10
alpha = 1
beta = 2
rho = 0.5
Q = 100
iterations = 100

# 4. Initialize pheromone
pheromone = [
    [1.0 if i != j else 0 for j in range(n)]
    for i in range(n)
]

best_route = None
best_distance = float('inf')

random.seed(42)

# 5. Main ACO algorithm
for iteration in range(iterations):

    all_routes = []

    for ant in range(num_ants):

        # Start from restaurant
        route = [0]
        unvisited = set(range(1, n))

        # Visit all customers
        while unvisited:

            current = route[-1]
            choices = list(unvisited)
            weights = []

            for city in choices:
                p = pheromone[current][city] ** alpha
                h = (1 / dist[current][city]) ** beta

                weights.append(p * h)

            # Select next customer
            next_city = random.choices(
                choices, weights=weights, k=1
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)

        # Return to restaurant
        length = 0

        for i in range(n - 1):
            length += dist[route[i]][route[i+1]]

        length += dist[route[-1]][route[0]]

        all_routes.append((route, length))

        # Update best route
        if length < best_distance:
            best_distance = length
            best_route = route.copy()

    # 6. Pheromone evaporation
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # 7. Pheromone deposit
    for route, length in all_routes:

        deposit = Q / length

        for i in range(n):
            a = route[i]
            b = route[(i + 1) % n]

            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

# 8. Display result
result = [names[i] for i in best_route]
result.append(names[0])

print("Best Delivery Route:")
print(" -> ".join(result))

print("Minimum Delivery Distance:",
      round(best_distance, 2))
```
