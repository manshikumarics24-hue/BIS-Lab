```python

import random
import math

# Evaluate Objective Function
def objective(x, expression):

    variables = {
        f"x{i+1}": x[i] for i in range(len(x))
    }

    functions = {
        "abs": abs,
        "sqrt": math.sqrt,
        "sin": math.sin,
        "cos": math.cos,
        "exp": math.exp,
        "log": math.log,
        "pi": math.pi
    }

    return eval(
        expression,
        {"__builtins__": {}},
        {**functions, **variables}
    )


# Tournament Selection
def selection(population, scores, maximize):

    candidates = random.sample(
        range(len(population)), 3
    )

    if maximize:
        winner = max(candidates, key=lambda i: scores[i])
    else:
        winner = min(candidates, key=lambda i: scores[i])

    return population[winner][:]


# Crossover
def crossover(parent1, parent2, crossover_rate):

    if random.random() < crossover_rate:

        child = []

        for i in range(len(parent1)):
            alpha = random.random()

            gene = (
                alpha * parent1[i] +
                (1 - alpha) * parent2[i]
            )

            child.append(gene)

        return child

    return parent1[:]


# Mutation
def mutation(child, mutation_rate, lower, upper):

    for i in range(len(child)):

        if random.random() < mutation_rate:

            child[i] = random.uniform(lower, upper)

    return child


# Main Genetic Algorithm
def genetic_algorithm():

    print("GENETIC ALGORITHM OPTIMIZATION")

    # Take input from user
    expression = input(
        "Enter objective function (use x1, x2...): "
    )

    choice = input("Enter max or min: ").lower()

    N = int(input("Enter population size: "))

    D = int(input("Enter number of variables: "))

    mutation_rate = float(
        input("Enter mutation rate (0 to 1): ")
    )

    crossover_rate = float(
        input("Enter crossover rate (0 to 1): ")
    )

    generations = int(
        input("Enter number of generations: ")
    )

    lower = float(input("Enter lower bound: "))

    upper = float(input("Enter upper bound: "))

    # Validate inputs
    if choice not in ["max", "min"]:
        print("Invalid optimization type")
        return

    if N < 3 or D < 1 or generations < 1:
        print("Invalid parameters")
        return

    if not (0 <= mutation_rate <= 1):
        print("Invalid mutation rate")
        return

    if not (0 <= crossover_rate <= 1):
        print("Invalid crossover rate")
        return

    if lower >= upper:
        print("Invalid bounds")
        return

    maximize = choice == "max"

    # Step 1: Create initial population
    population = [
        [random.uniform(lower, upper) for _ in range(D)]
        for _ in range(N)
    ]

    # Fitness evaluation
    def evaluate(individual):
        return objective(individual, expression)

    # Step 2: Evolution process
    for generation in range(generations):

        scores = [evaluate(ind) for ind in population]

        # Find best individual
        best_index = (
            max(range(N), key=lambda i: scores[i])
            if maximize
            else min(range(N), key=lambda i: scores[i])
        )

        best_individual = population[best_index][:]

        # Step 3: Elitism
        new_population = [best_individual]

        # Step 4: Selection, Crossover and Mutation
        while len(new_population) < N:

            parent1 = selection(
                population, scores, maximize
            )

            parent2 = selection(
                population, scores, maximize
            )

            # Generate offspring
            child = crossover(
                parent1, parent2, crossover_rate
            )

            # Apply mutation
            child = mutation(
                child, mutation_rate, lower, upper
            )

            new_population.append(child)

        population = new_population

    # Step 5: Final fitness evaluation
    scores = [evaluate(ind) for ind in population]

    best_index = (
        max(range(N), key=lambda i: scores[i])
        if maximize
        else min(range(N), key=lambda i: scores[i])
    )

    best_solution = population[best_index]
    best_fitness = scores[best_index]

    # Step 6: Display result
    print("\n----- FINAL RESULT -----")

    print("Best Solution:", best_solution)

    print("Optimal Fitness Value:", best_fitness)


# Execute Algorithm
genetic_algorithm()
```
